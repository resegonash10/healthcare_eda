# Section 4 – Data Cleaning

## Objective

After checking the quality of the dataset, the next step is to clean the data.

The purpose of data cleaning is to improve the quality of the dataset before performing any analysis. In this section, duplicate records will be removed, date columns will be converted into the correct format, negative billing amounts will be handled, and patient names will be standardised.

### Step 1 – Create a Copy of the Dataset

To preserve the original dataset, all cleaning will be performed on a copy.

```python
# create a copy of the original dataset
df_clean = df.copy()
```

Explanation

We never clean the original dataset directly.

If something goes wrong, we can always go back to the original dataframe.

Step 2 – Remove Duplicate Rows

### Step 2 – Remove Duplicate Rows

Check the size of the dataset before and after removing duplicate records.

```python
# shape after removing duplicates
df_clean.shape
```

### Interpretation

Duplicate records were removed from the dataset. Comparing the dataset shape before and after cleaning confirms whether any duplicate rows were found.

### Step 3 – Convert Date Columns

Convert the admission and discharge dates from text format into datetime format.

```python
# convert date of admission to datetime
# Convert Date of Admission to datetime

df_clean["Date of Admission"] = pd.to_datetime(df_clean["Date of Admission"])

df_clean["Discharge Date"] = pd.to_datetime(df_clean["Discharge Date"])
```

```python
# check the data types
print(df_clean.dtypes)
```

The Date of Admission and Discharge Date columns were successfully converted into datetime format. This allows date calculations and feature engineering to be performed later in the analysis.

### Step 4 – Handle Negative Billing Amounts

Negative billing amounts are not realistic for hospital charges. These records will be removed from the dataset.

```python
# Remove rows with negative billing amount
df_clean = df_clean[df_clean["Billing Amount"] >= 0]
```

### Interpretation

Rows containing negative billing amounts were removed because hospital bills are expected to have positive values. Removing these records improves the reliability of the dataset.


### Step 5 – Standardise Patient Names

Convert patient names into a consistent format.

### Interpretation

Patient names were converted into title case to ensure consistent formatting throughout the dataset.

## Cleaning Summary

The data cleaning process was performed on a copy of the original dataset to preserve the raw data.

# Section 5 – Feature Engineering

## Objective

Feature engineering is the process of creating new variables from existing data to make the dataset more useful for analysis.

In this section, eight new columns will be created using the cleaned dataset. These new columns will help answer the analysis questions in the next section.

### Step 1 – Create the Length of Stay Column

Calculate the number of days each patient stayed in hospital.

```python
# Calculate the length of stay
df_clean["Length_of_Stay"] = (
    df_clean["Discharge Date"] - df_clean["Date of Admission"]
).dt.days
```

### Interpretation

The Length_of_Stay column was successfully created and shows the total number of days each patient spent in hospital.

### Step 2 – Create the Admission Year Column

Extract the year from the admission date.

```python
# Extract admission year
df_clean["Admission_Year"] = df_clean["Date of Admission"].dt.year
```

### Interpretation

The Admission_Year column was created to make it easier to analyse admissions over different years.


### Step 3 – Create the Admission Month Column

Extract the month number from the admission date.

```python
# Extract admission month
df_clean["Admission_Month"] = df_clean["Date of Admission"] .dt.month
```

### Interpretation

The Admission_Month column stores the month as a number from 1 to 12.

### Step 4 – Create the Admission Month Name Column

Extract the month name from the admission date.

```python
# Extract month name
df_clean['Admission_Month_Name'] = (df_clean["Date of Admission"].dt.month_name())
```

### Interpretation

The Admission_Month_Name column stores the full month name, making reports easier to read.

### Step 5 – Create the Admission Day Name Column

Extract the weekday from the admission date.

```python
# Extract weekday name
df_clean["Admission_Day_Name"] = (df_clean["Date of Admission"].dt.day_name())
```

### Interpretation

The Admission_Day_Name column was created to analyse which days of the week patients are admitted.

### Step 6 – Create the Age Group Column

Group patients into age categories using `.loc`.

```python
# Create a default valve
df_clean["Age_Group"] = ""

# Child (0–17)

df_clean.loc[df_clean["Age"] <= 17, "Age_Group"] = "Child"

# Young Adult (18–35)

df_clean.loc[
    (df_clean["Age"] >= 18) & (df_clean["Age"] <= 35),
    "Age_Group"
] = "Young Adult"

# Adult (36–50)

df_clean.loc[
    (df_clean["Age"] >= 36) & (df_clean["Age"] <= 50),
    "Age_Group"
] = "Adult"# Senior (51–65)

df_clean.loc[
    (df_clean["Age"] >= 51) & (df_clean["Age"] <= 65),
    "Age_Group"
] = "Senior"

# Elderly (66+)

df_clean.loc[df_clean["Age"] >= 66, "Age_Group"] = "Elderly"
```

### Interpretation

Patients were grouped into age categories to simplify comparisons during the analysis stage.

### Step 7 – Create the Billing Bucket Column

Group billing amounts into cost categories using `.loc`.

```python
# Create default value

df_clean["Billing_Bucket"] = ""

# Low

df_clean.loc[
    df_clean["Billing Amount"] < 10000,
    "Billing_Bucket"
] = "Low"

# Medium

df_clean.loc[
    (df_clean["Billing Amount"] >= 10000) &
    (df_clean["Billing Amount"] < 20000),
    "Billing_Bucket"
] = "Medium"

# High

df_clean.loc[
    (df_clean["Billing Amount"] >= 20000) &
    (df_clean["Billing Amount"] < 30000),
    "Billing_Bucket"
] = "High"

# Very High

df_clean.loc[
    df_clean["Billing Amount"] >= 30000,
    "Billing_Bucket"
] = "Very High"

```

### Interpretation

Billing amounts were grouped into four categories to simplify cost analysis.


### Step 8 – Create the Stay Category Column

Group patients according to the number of days spent in hospital.


```python
# Create default value

df_clean["Stay_Category"] = ""

# Short Stay

df_clean.loc[
    df_clean["Length_of_Stay"] <= 3,
    "Stay_Category"
] = "Short"

# Medium Stay

df_clean.loc[
    (df_clean["Length_of_Stay"] >= 4) &
    (df_clean["Length_of_Stay"] <= 7),
    "Stay_Category"
] = "Medium"

# Long Stay

df_clean.loc[
    df_clean["Length_of_Stay"] >= 8,
    "Stay_Category"
] = "Long"
```

### Interpretation

Patients were grouped into short, medium, and long hospital stays to make it easier to compare admission patterns.

### Step 9 – Display the New Columns

Preview the cleaned dataset to confirm that all new columns were created successfully.

```python
# Display the first five row
print(df_clean.head())
```

## Feature Engineering Summary

Eight new columns were successfully created from the existing data.

Five date-related columns were created using the admission and discharge dates, while three additional grouping columns were created using the `.loc` method as required by the assignment.

These new features will make the analysis in the next section easier to perform and interpret.
